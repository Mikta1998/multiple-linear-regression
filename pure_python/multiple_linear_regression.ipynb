```python
%matplotlib widget
from pure_python.functions import mean_squared_error
import numpy as np
import matplotlib.pyplot as plt
import random
```

```python
X = [
    [1, 2],
    [2, 1],
    [3, 4],
    [4, 3],
    [5, 6],
    [6, 5],
    [7, 3],
    [8, 7],
    [9, 5],
    [10, 8]
]

y_intercept = random.random()
slope_values = [random.random() for _ in range(len(X[0]))]

y_true = [10, 7, 21, 17, 30, 26, 27, 36, 37, 43]
y_predicted = []

for feature_number in range(len(X)):
    row_prediction = []
    for slope_value in range(len(slope_values)):
        row_prediction.append(slope_values[slope_value] * X[feature_number][slope_value])
    y_predicted.append(y_intercept + sum(row_prediction))

print(y_true)
print(y_predicted)
```

```python
for feature_values in range(len(X)):
    print(f" First Feature: {X[feature_values][0]} Second Feature: {X[feature_values][1]}, Slope Value 1: {slope_values[0]}, Slope Value 2: {slope_values[1]} Y-Intercept: {y_intercept}")
```

```python
print(f"Loss: {mean_squared_error(y_true, y_predicted)}")
```

```python
x_1 = []
x_2 = []
z_values = []
for feature_values in range(len(X)):
    x_1.append(X[feature_values][0])
    x_2.append(X[feature_values][1])
```

```python
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection="3d")

ax.plot_trisurf(x_1, x_2, y_true)

ax.plot_trisurf(x_1, x_2, y_predicted, color="red")

ax.set_xlabel("Feature 1")
ax.set_ylabel("Feature 2")
ax.set_zlabel("Y-Intercept")

plt.show()
```
