```python
%matplotlib widget
from pure_python.functions import mean_squared_error, calculate_intercept_gradient, calculate_slope_gradient, \
    update_intercept, update_slope, LEARNING_RATE
import numpy as np
import matplotlib.pyplot as plt
import random
```

```python
X = [
    [1, 2],
    [2, 1],
    [3, 4],
    [4, 3],
    [5, 6],
    [6, 5],
    [7, 3],
    [8, 7],
    [9, 5],
    [10, 8]
]

y_true = [10, 7, 21, 17, 30, 26, 27, 36, 37, 43]
```

```python
ITERATIONS = 100

slopes = []
intercept = []
loss_history = []
prediction_history = {}

y_intercept = random.random()
slope_values = [random.random() for _ in range(len(X[0]))]

slopes.append(slope_values)
intercept.append(y_intercept)

for iteration in range(ITERATIONS):
    y_predicted = []
    for feature_number in range(len(X)):
        row_prediction = []
        for slope_value in range(len(slope_values)):
            row_prediction.append(slope_values[slope_value] * X[feature_number][slope_value])
        y_predicted.append(y_intercept + sum(row_prediction))

    prediction_history[iteration] = y_predicted.copy()

    mse = mean_squared_error(y_true, y_predicted)
    loss_history.append(mse)

    y_intercept_new = update_intercept(X, y_true, slope_values, y_intercept)
    slope_values_new = update_slope(X, y_true, slope_values, y_intercept)

    y_intercept = y_intercept_new
    slope_values = slope_values_new

    slopes.append(slope_values_new)
    intercept.append(y_intercept)

    print(f" Iteration: {iteration} Slope Value 1: {slope_values[0]}, Slope Value 2: {slope_values[1]} Y-Intercept: {y_intercept} Loss: {loss_history[-1]}")

prediction_history[ITERATIONS] = y_predicted.copy()

y_predicted = []
for feature_number in range(len(X)):
    row_prediction = []
    for slope_value in range(len(slope_values)):
        row_prediction.append(slope_values[slope_value] * X[feature_number][slope_value])
    y_predicted.append(y_intercept + sum(row_prediction))

mse = mean_squared_error(y_true, y_predicted)
loss_history.append(mse)

print(f"Iteration: {ITERATIONS} Slope Value 1: {slope_values[0]}, Slope Value 2: {slope_values[1]} Y-Intercept: {y_intercept} Loss: {loss_history[-1]}")
```
